## Part 7: The Line of Best Fit (Manual Regression)

**Objective:** Intuition for "Weights" and "Bias" using simple pricing.

Before we classify types, let's try to predict the toll price based on weight.

$$\text{Price} = (m \times \text{Weight}) + c$$

1. We assume the relationship is linear: `Price = (m * Weight) + c`.
2. **Visualize & Minimize:** Run the helper function `visualize_regressor(m, c)`.
3. **Manual Tuning:** Start with `m = 0`, `c = 0`. Note the high error score. Adjust `m` (slope) and `c` (intercept) until the line passes through the data and the Error Metric is minimized (closest to 0).
4. **Concept Check:** What does `m` represent in the real world? (Hint: Cost per ton).


```python
def visualize_regressor(m, c):
    """Visualize y = mx + c against a simple pricing dataset."""
    weights = np.array([1, 2, 5, 8, 10])
    actual_prices = np.array([15, 25, 55, 85, 105])

    predicted_prices = (m * weights) + c
    avg_error = np.mean(np.abs(actual_prices - predicted_prices))

    plt.figure(figsize=(8, 5))
    plt.scatter(weights, actual_prices, color="green", s=100, label="Actual Data")
    plt.plot(weights, predicted_prices, color="purple", linewidth=3, label="Your Line")

    plt.xlabel("Weight (tons)")
    plt.ylabel("Toll Price ($)")
    plt.title(f"Linear Regression Tuning | Average Error: ${avg_error:.2f}")
    plt.legend()
    plt.grid(True)
    plt.show()

    print(f"Average Error: ${avg_error:.2f}")
```
